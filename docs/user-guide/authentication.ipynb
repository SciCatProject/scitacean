# Authentication



## SciCat


### Token


````python
from scitacean import Client

client = Client.from_token("ess", token="my-token")
````

````python
from scitacean import Client
from scitacean.transfer.sftp import SFTPFileTransfer

client = Client.from_token(
    url="https://scicat.example.com/api/v3",
    file_transfer=SFTPFileTransfer(host="example.com"),
    token="my-token",
)
````

### Password

````python
from scitacean import Client

client = Client.from_credentials(
    "ess",
    username="my-user",
    password="my-password",
)
````

````python
from scitacean import Client
from scitacean.transfer.sftp import SFTPFileTransfer

client = Client.from_credentials(
    url="https://scicat.example.com/api/v3",
    file_transfer=SFTPFileTransfer(host="example.com"),
    username="my-user",
    password="my-password",
)
````

## File transfer

### SSH / SFTP

````python
from scitacean.transfer.sftp import SFTPFileTransfer

transfer = SFTPFileTransfer(host="example.com")
client = Client.without_login(
    url="https://scicat.example.com/api/v3",
    file_transfer=transfer,
)
````

````python
from scitacean.transfer.sftp import SFTPFileTransfer

transfer = SFTPFileTransfer(
    host="example.com",
    key_filename="/home/my-user/.ssh/scicat-key",
)
````

````python
from scitacean.transfer.sftp import SFTPFileTransfer

transfer = SFTPFileTransfer(
    host="example.com",
    username="my-user",
    password="my-password",
)
````

connect is described also in docstring of SFTPFileTransfer.

connection only opened on demand and closed asap

````python
from paramiko import SFTPClient, SSHClient
from scitacean.transfer.sftp import SFTPFileTransfer


def connect(host: str, port: int) -> SFTPClient:
    ssh_client = SSHClient()
    ssh_client.load_system_host_keys()
    ssh_client.connect(
        hostname=host,
        port=port,
        username="my-username",
        key_filename="/home/my-user/.ssh/scicat-key",
    )
    return ssh_client.open_sftp()


transfer = SFTPFileTransfer(
    host="example.com",
    connect=connect,
)
````
